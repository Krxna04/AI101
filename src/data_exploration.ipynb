```python
#IMPORTING libaries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Plot settings
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 5)

# Display more columns when inspecting the dataset
pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", lambda x: f"{x:,.4f}")
```

```python
#Getting Data
DATA_PATH = "creditcard.csv"

df = pd.read_csv(DATA_PATH)

print("Dataset loaded successfully!")
print("Shape (rows, columns):", df.shape)
display(df.head())

print("Column names:")
print(df.columns.tolist())

print("\\nDataset information:")
df.info()
print("Numerical summary:")
display(df.describe().T)
```

```python
#Collecting missing values
missing = df.isnull().sum().sort_values(ascending=False)
missing_summary = pd.DataFrame({
    "Missing Values": missing,
    "Missing Percentage": (missing / len(df) * 100).round(2)
})
display(missing_summary[missing_summary["Missing Values"] > 0])

print("Total missing values:", int(df.isnull().sum().sum()))
print("Duplicate rows:", int(df.duplicated().sum()))
```

```python
#Explore data distribution
class_counts = df["Class"].value_counts().sort_index()
class_percentages = df["Class"].value_counts(normalize=True).sort_index() * 100

class_summary = pd.DataFrame({
    "Transaction Count": class_counts,
    "Percentage (%)": class_percentages.round(4)
})
display(class_summary)

plt.figure(figsize=(7, 5))
ax = sns.countplot(data=df, x="Class")
ax.set_title("Distribution of Legitimate and Fraudulent Transactions")
ax.set_xlabel("Class (0 = Legitimate, 1 = Fraud)")
ax.set_ylabel("Number of Transactions")
plt.tight_layout()
plt.savefig(
    "../output/graphs/class_distribution.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()

legitimate = int((df["Class"] == 0).sum())
fraudulent = int((df["Class"] == 1).sum())

print(f"Legitimate transactions: {legitimate:,}")
print(f"Fraudulent transactions: {fraudulent:,}")
if fraudulent:
    print(f"Legitimate-to-fraud ratio: {legitimate / fraudulent:,.2f} : 1")
print(f"Fraud percentage: {fraudulent / len(df) * 100:.4f}%")
```

```python
#Compare thr transaction features
print("Transaction Amount Summary:")
display(df["Amount"].describe())

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.histplot(df["Amount"], bins=50, ax=axes[0])
axes[0].set_title("Transaction Amount Distribution")
axes[0].set_xlabel("Amount")

sns.boxplot(x=df["Amount"], ax=axes[1])
axes[1].set_title("Transaction Amount Boxplot")
axes[1].set_xlabel("Amount")
plt.savefig(
    "../output/graphs/Transaction Amount Distribution.png",
    dpi=300,
    bbox_inches="tight"
)
plt.tight_layout()
plt.show()

plt.figure(figsize=(9, 5))
sns.histplot(np.log1p(df["Amount"]), bins=50)
plt.title("Log-Transformed Transaction Amount Distribution")
plt.xlabel("log(1 + Amount)")
plt.tight_layout()
plt.savefig(
    "../output/graphs/transaction_distribution.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()

display(
    df.groupby("Class")["Amount"]
      .agg(["count", "mean", "median", "min", "max", "std"])
      .rename(index={0: "Legitimate (0)", 1: "Fraud (1)"})
)

plt.figure(figsize=(9, 5))
sns.boxplot(data=df, x="Class", y="Amount", showfliers=False)
plt.title("Transaction Amount by Class (outliers hidden)")
plt.xlabel("Class (0 = Legitimate, 1 = Fraud)")
plt.ylabel("Amount")
plt.tight_layout()
plt.savefig(
    "../output/graphs/transaction_byClass.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
```

```python
#Exploring Time
if "Time" in df.columns:
    plt.figure(figsize=(10, 5))
    sns.histplot(data=df, x="Time", hue="Class", bins=50, element="step",
                 stat="density", common_norm=False)
    plt.title("Transaction Time Distribution by Class")
    plt.xlabel("Seconds elapsed from the first transaction")
    plt.tight_layout()
    plt.savefig(
        "../output/graphs/transaction_time.png",
        dpi=300,
        bbox_inches="tight"
    )
    plt.show()
else:
    print("The Time column is not present in this dataset.")
```

```python
# Correlation matrix
corr = df.corr(numeric_only=True)

plt.figure(figsize=(16, 12))
sns.heatmap(corr, cmap="coolwarm", center=0, linewidths=0.1)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.savefig(
    "../output/graphs/corelation_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
```

```python
# The anonymized features are named V1, V2, ... in the common credit-card dataset.
v_features = [col for col in df.columns if col.startswith("V")]

if v_features:
    selected_features = v_features[:6]  # Plot a manageable subset first
    fig, axes = plt.subplots(2, 3, figsize=(16, 9))
    axes = axes.flatten()

    for ax, feature in zip(axes, selected_features):
        sns.kdeplot(data=df, x=feature, hue="Class", common_norm=False,
                    fill=False, ax=ax)
        ax.set_title(f"{feature} distribution by class")

    plt.tight_layout()
    plt.savefig(
        "../output/graphs/feature_distributon.png",
        dpi=300,
        bbox_inches="tight"
    )
    plt.show()
else:
    print("No V1, V2, ... features found.")
```
