```python
#Executing notebooks
from IPython import get_ipython

ip = get_ipython()

def run_notebook(notebook_name, notebook_file):

    print("\n" + "=" * 65)
    print("EXECUTING:", notebook_name)
    print("=" * 65)

    notebook_path = PROJECT_DIR / notebook_file

    nb = nbformat.read(
        notebook_path,
        as_version=4
    )

    # Execute every code cell in the current kernel
    for index, cell in enumerate(nb.cells):

        if cell.cell_type == "code":

            code = cell.source.strip()

            if not code:
                continue

            print(
                f"\nRunning cell {index + 1} "
                f"from {notebook_file}"
            )

            result = ip.run_cell(
                code,
                store_history=False
            )

            if result.error_before_exec:
                raise result.error_before_exec

            if result.error_in_exec:
                raise result.error_in_exec

        print("\nCompleted:", notebook_name)

run_notebook(
    "Data Exploration",
    NOTEBOOKS["Data Exploration"]
)

run_notebook(
    "Data Preprocessing",
    NOTEBOOKS["Data Preprocessing"]
)

required_variables = [
    "X_train",
    "X_test",
    "y_train",
    "y_test"
]

for variable in required_variables:

    if variable not in globals():

        raise NameError(
            f"{variable} was not created "
            "by the preprocessing notebook."
        )

    print(variable, "created successfully!")

print("\nTraining data shape:", X_train.shape)
print("Testing data shape:", X_test.shape)

run_notebook(
    "Model Training",
    NOTEBOOKS["Model Training"]
)

model = Pipeline

if "model" not in globals():

    raise NameError(
        "Model not found! Check the model training notebook."
    )

print("Model trained successfully!")

print("\nModel Details:")
print(model)

print("\nModel Type:")
print(type(model))

run_notebook(
    "Model Evaluation",
    NOTEBOOKS["Model Evaluation"]
)
```
